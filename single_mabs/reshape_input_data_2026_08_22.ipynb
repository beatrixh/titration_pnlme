```python
path="input_data/retagged_data/revised_subset_full_plates_some_neutralization_2026-08-10_RETAGGED_2026_08_17.csv"
df = pd.read_csv(path)
```

```python
wide
```

```python
vcs = df.loc[df.specrole=='vc']
ccs = df.loc[df.specrole=='cc']

vcs = vcs.pivot(
    index='run_id',
    columns='plate_row',
    values='rlu'
)

vcs.columns = [f"vc_{i}" for i in vcs.columns]

ccs = ccs.pivot(
    index='run_id',
    columns='plate_row',
    values='rlu'
)

ccs.columns = [f"cc_{i}" for i in ccs.columns]
```

```python
samples = df.loc[df.specrole=='sample']
samples = samples.merge(vcs.reset_index(), on='run_id')
samples = samples.merge(ccs.reset_index(), on='run_id')
```
